```python
import json

with open("final_bigram.json", "r") as f:
    bigram_counts = json.load(f)  # Charge tout si c'est un seul gros dictionnaire
```

```python
bigram_probs = {}

for w1, followers in bigram_counts.items():
    total_count = sum(followers.values())  # Nombre total d'occurrences du mot w1
    bigram_probs[w1] = {w2: count / total_count for w2, count in followers.items()} 
```

```python
with open("bigram_probs2.json","w") as f:
    json.dump(bigram_probs,f,indent=4)
```
